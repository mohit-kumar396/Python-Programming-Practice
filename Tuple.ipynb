```python
#Method 1:- tuple Unpacking
def get_min_max_sum(numbers):
  return min(numbers), max(numbers), sum(numbers)

minimum,maximum,sum=get_min_max_sum((15, 25, 10, 40, 35, 20))
print(min)
print(max)
print(sum)
```

```python
# Method 2:- Write the logic manually:-
# Write the logic manually:-

def get_min_max_sum(numbers):
  minimum = numbers[0]
  maximum = numbers[0]
  total = 0
  for num in numbers:
    if num<minimum:
      minimum=num

    if num>maximum:
      maximum=num

    total += num
  return minimum,maximum,total

```

```python
minimum,maximum,total=get_min_max_sum((15, 25, 10, 40, 35, 20))
print(minimum)
print(maximum)
print(total)
```

#**Q29. Flatten a nested tuple:**

```python
# Write a Python program to flatten the following nested tuple into a single tuple:
# nested = (1, (2, 3), (4, (5, (6, 7))))
# Expected result:
# (1, 2, 3, 4, 5, 6, 7)
```

```python
#Method1:- Hardcoded Unpacking
a,b,c=nested
```

```python
print(a)
print(b)
print(c)
```

```python
b1,b2=b
```

```python
c1,d=c
```

```python
d1,d2=d
```

```python
e1,e2=d2
```

```python
final_flatten_tupple=(a,b1,b2,c1,d1,e1,e2)
print(final_flatten_tupple)
```

```python
# Method 2:-

def flatten(t):
  result=()

  for item in t:
    if isinstance(item,tuple):
      result+=flatten(item)

    else:
      result += (item, )
  return result
print(flatten(nested))
```

#**Q30. Compare memory usage:**

```python
# Write a Python program using the sys module to create a list and a tuple containing one
# million integers. Compare and display the memory consumed by the list and tuple
```

```python
import sys

lst = list(range(1_000_000)) # Important concept shortcut hai direct 0 to 999999 tak ka list bna dega
tup = tuple(range(1_000_000)) # Important concept shortcut hai direct 0 to 999999 tak ka tuple bna dega

print("List memory:", sys.getsizeof(lst), "bytes")
print("Tuple memory:", sys.getsizeof(tup), "bytes")

# Notes:- sys.getsizeof() se object ka memory size check karte hai.
```
