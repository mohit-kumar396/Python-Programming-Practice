```python
# 10. Write a Python program to create a dictionary containing student names and marks, and display
# the name and marks of the student who has the highest marks.
```

```python
student = {
    "Rahul": 85,
    "Aman": 92,
    "Priya": 78,
    "Rohit": 95,
    "Neha": 88
}

highest_student = max(student, key=student.get)
print("Student:", highest_student)
print("Marks:", student[highest_student])
```

```python
# 40. Write a Python program to create a nested dictionary containing student names and their marks
# in three subjects. Calculate and display the total marks and average marks of each student.
```

```python
students = {
    "Rahul": {
        "Python": 85,
        "SQL": 90,
        "Math": 80
    },
    "Aman": {
        "Python": 78,
        "SQL": 88,
        "Math": 92
    },
    "Priya": {
        "Python": 90,
        "SQL": 85,
        "Math": 95
    }
}

for name, marks in students.items():
    total = marks["Python"] + marks["SQL"] + marks["Math"]
    average = total / 3

    print("Student:", name)
    print("Total marks:", total)
    print("Average marks:", average)
    print()
```
