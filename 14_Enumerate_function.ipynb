```python
# ============================================================
# enumerate()
# ============================================================

# Definition:
# enumerate() is used to get both the index and value
# of each element while iterating over an iterable.


# Syntax:
#enumerate(iterable)

# OR

#for index, value in enumerate(iterable):
    # code


# Basic Example:

names = ["Rahul", "Aman", "Mohit"]

for index, value in enumerate(names):
  print(index, value)

# Output:
# 0 Rahul
# 1 Aman
# 2 Mohit


# Explanation:
# index  -> position/index of the current element
# value  -> current element
# iterable -> list, tuple, string, etc.
#
# enumerate() gives:
# (index, value)
#
# Example:
# (0, "Rahul")
# (1, "Aman")
# (2, "Mohit")


# Starting index from a custom number:

for index, value in enumerate(names, start=1):
  print(index,value)

# Output:
# 1 Rahul
# 2 Aman
# 3 Mohit


# Important:
# By default, enumerate() starts indexing from 0.
# We can change it using start.


# ------------------------------------------------------------
# Simple way to remember:
# ------------------------------------------------------------
# enumerate() = index + value
#
# enumerate(iterable)
#        ↓
# (index, value)
```
