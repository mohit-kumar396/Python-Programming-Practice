```python
a=list(map(int, input("Enter the values in a:").split()))
```

```python
a
```

```python
cum_sum=[]
for i in range(0,len(a)):
  total=0
  for j in range(0,i+1):
    total=total+a[j]
  cum_sum.append(total)
```
