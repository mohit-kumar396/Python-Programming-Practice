```python
#Method 1:-
lst = list(map(int, input("Enter the numbers: ").split()))

lst.sort()

for i in range(len(lst)):
    if i == 0 or lst[i] != lst[i-1]:
        track = lst[i]
        count = 0

        for j in lst:
            if j == track:
                count += 1

        print(track, count)
```

```python
# method 2:- always reliable for count frequencies

lst = list(map(int, input("Enter the numbers: ").split()))

freq = {}

for i in lst:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1

print(freq)
```
