```python
# Normal list comprehension:-condition ka result add karo → True/False
# Syntax :-
# [expression for item in iterable]
# it will return array with true or false based on condition
```

```python
# Filtering: sirf condition satisfy karne wale i add karo
# [expression for item in iterable if condition]
```

```python
# Q1 :- create a new list containing element greater than 20
```

```python
result=[i for i in range(30) if i>20]
result
```

```python
# Q2 :- Using list comprehension print all even element of list.
```

```python
lst=list(map(int, input("Enter the value in list").split()))
```

```python
result2=[lst[i] for i in range(len(lst)) if lst[i]%2==0]
result2
```
