```python
# Normal list comprehension:-condition ka result add karo → True/False
# Syntax :-
# [expression for item in iterable]
# it will return array with true or false based on condition
```

```python
# Filtering: sirf condition satisfy karne wale i add karo
# [expression for item in iterable if condition]
```

```python
# Q1 :- create a new list containing element greater than 20
```

```python
result=[i for i in range(30) if i>20]
result
```

```python
# Q2 :- Using list comprehension print all even element of list.
```

```python
lst=list(map(int, input("Enter the value in list").split()))
```

```python
result2=[lst[i] for i in range(len(lst)) if lst[i]%2==0]
result2
```

```python
# Q3:- Doubble all even numbers of list while keeping the odd remain unchanged.
# Note :- if you want to tranform array use if else befor
```

```python
lst2=[5, 10, 15, 20, 25, 30]
```

```python
[lst2[i]*2 if lst2[i]%2==0 else lst2[i] for i in range(len(lst2))]
```

```python
# Q4:- Doubble all even numbers of list and return them
# Here we will use if at end because we are filtering
```

```python
[x*2 for x in lst2 if x%2==0]
```

```python
# Note :- List comprehension creates a new list; it does not modify the original
# list unless you assign the result back to the original variable.
# thats why Q4 ka answer [20, 40, 60] ye aya na ki [40,80,120]
```
