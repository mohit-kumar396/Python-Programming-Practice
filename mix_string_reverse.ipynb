```python
s1=input("Enter the value in string 1: ")
```

```python
s2=input("Enter the value in string 1: ")
```

```python
#List Reverse
rev_s2=s2[::-1]
```

```python
if len(s1)>len(s2):
  ratio=len(s1)
else:
  ratio=len(s2)
```

```python
result=""
for i in range(ratio):
  if i<len(s1):
    result+=s1[i]
  if i<len(rev_s2):
    result+=rev_s2[i]
```
