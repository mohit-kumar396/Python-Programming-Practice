#**Q19 Set of Even and Odd Numbers**

```python
# Create a set containing the numbers from 1 to 20. Use set operations or appropriate logic
# to create separate sets containing the even numbers and odd numbers.
```

```python
num=set()
for i in range(1,21):
  num.add(i)
num
```

```python
even=set()
odd=set()
for i in num:
  if i%2==0:
    even.add(i)
  else:
    odd.add(i)
even
```
