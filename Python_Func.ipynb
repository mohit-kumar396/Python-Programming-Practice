````python
# 3. Write a Python function that accepts three numbers as arguments and returns the maximum of
# the three numbers.
````

````python
def find_max_min(a, b, c):
    if a > b and a > c:
        maximum = a
    elif b > a and b > c:
        maximum = b
    else:
        maximum = c

    if a < b and a < c:
        minimum = a
    elif b < a and b < c:
        minimum = b
    else:
        minimum = c

    return maximum, minimum


maximum, minimum = find_max_min(10, 25, 5)

print("Maximum:", maximum)
print("Minimum:", minimum)
````

````python
# 4. Write a Python function that accepts list as arguments and returns the maximum and minimum.
````

````python
def find_max_min(numbers):
    maximum = numbers[0]
    minimum = numbers[0]

    for num in numbers:
        if num > maximum:
            maximum = num

        if num < minimum:
            minimum = num

    return maximum, minimum


numbers = [10, 25, 5, 40, 15]

maximum, minimum = find_max_min(numbers)

print("Maximum:", maximum)
print("Minimum:", minimum)
````

````python
# 5. Write a Python function that accepts a word as an argument and counts the number of vowels
# and consonants in the word.
````

````python
def count_vowels_consonants(word):
    vowel_count = 0
    consonant_count = 0

    for char in word.lower():
        if char=='a' or char=='e' or char=='i' or char=='o'or char=='u' :
            vowel_count += 1
        else:
            consonant_count += 1

    return vowel_count, consonant_count


word = "PythOn"

vowel_count, consonant_count = count_vowels_consonants(word)

print("Vowels:", vowel_count)
print("Consonants:",consonant_count )
````

````python
# 6. Write a Python function that accepts a positive integer as an argument and returns its factorial.
````

````python
# Method 1. Normal method — for loop

def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result


number = int(input("Enter the number you want factorial of:- "))

result = factorial(number)

print("Factorial:", result)
````

````python
# Method 2. Recursion method

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


number = 5

result = factorial(number)

print("Factorial:", result)
````

````python
# # Python `*args`

# ## What is `*args`?

# `*args` is used in a Python function when we **don't know how many positional arguments** will be passed to the function.

# ### Basic Syntax

# ```python
# def function_name(*args):
#     # code
# ```

# Here, `args` is just a variable name. We can technically use another name, but `args` is the standard convention.

# ---

# ## Why do we use `*args`?

# Suppose we have a normal function:

# ```python
# def add(a, b):
#     return a + b
# ```

# This function accepts only **2 arguments**:

# ```python
# add(10, 20)       # Works
# add(10, 20, 30)   # Error
# ```

# But with `*args`:

# ```python
# def add(*args):
#     # any number of arguments can be accepted
# ```

# Now we can do:

# ```python
# add(10)
# add(10, 20)
# add(10, 20, 30)
# add(10, 20, 30, 40, 50)
# ```

# ---

# ## Important: `args` becomes a Tuple

# When we pass multiple values using `*args`, Python collects them into a **tuple**.

# Example:

# ```python
# def show(*args):
#     print(args)

# show(10, 20, 30, 40)
# ```

# Output:

# ```text
# (10, 20, 30, 40)
# ```

# So internally:

# ```python
# args = (10, 20, 30, 40)
# ```

# Therefore, we can use a `for` loop on `args`:

# ```python
# def show(*args):
#     for number in args:
#         print(number)

# show(10, 20, 30)
# ```

# Output:

# ```text
# 10
# 20
# 30
# ```

# ---

# # Example: Sum of Any Number of Numbers

# ```python
# def calculate_sum(*args):
#     total = 0

#     for number in args:
#         total += number

#     return total


# result = calculate_sum(10, 20, 30, 40)

# print("Sum:", result)
# ```

# Output:

# ```text
# Sum: 100
# ```

# ### How it works

# When we call:

# ```python
# calculate_sum(10, 20, 30, 40)
# ```

# Python collects all values:

# ```python
# args = (10, 20, 30, 40)
# ```

# Then:

# ```text
# total = 0

# 10 → total = 10
# 20 → total = 30
# 30 → total = 60
# 40 → total = 100
# ```

# Finally:

# ```python
# return total
# ```

# returns:

# ```text
# 100
# ```

# ---

# ## Another Example

# ```python
# def calculate_sum(*args):
#     total = 0

#     for number in args:
#         total += number

#     return total


# print(calculate_sum(5, 10))
# print(calculate_sum(5, 10, 15))
# print(calculate_sum(5, 10, 15, 20, 25))
# ```

# Output:

# ```text
# 15
# 30
# 75
# ```

# Notice that the same function can accept **different numbers of arguments**.

# ---

# ## ⭐ Remember This

# ```python
# def function_name(*args):
# ```

# means:

# > **"Accept any number of positional arguments and collect them inside `args` as a tuple."**

# ### Quick Memory Trick

# **`*args` → many positional arguments → stored as tuple**

# Example:

# ```python
# function(10, 20, 30, 40)
# ```

# becomes:

# ```python
# args = (10, 20, 30, 40)
# ```
````

````python
# 9. Write a Python function using *args to accept any number of numbers and return their sum.
````

````python
def calculate_sum(*args):
    total = 0

    for number in args:
        total += number

    return total


n = int(input("How many numbers do you want to enter? "))

numbers = []

for i in range(n):
    number = int(input("Enter number: "))
    numbers.append(number)

result = calculate_sum(*numbers)

print("Sum:", result)
````

````python
def find_max_min(*args):
    maximum = args[0]
    minimum = args[0]

    for number in args:
        if number > maximum:
            maximum = number

        if number < minimum:
            minimum = number

    return maximum, minimum


maximum, minimum = find_max_min(10, 25, 5, 40, 15)

print("Maximum:", maximum)
print("Minimum:", minimum)
````

````python
# 12. Write a Python function using *args to accept multiple strings and display the longest string.
````

````python
def longest_string(*args):
    longest = args[0]

    for word in args:
        if len(word) > len(longest):
            longest = word

    return longest


result = longest_string("Python", "Data", "MachineLearning", "SQL")

print("Longest string:", result)
````

````python
# 19. Write a Python function using both *args and **kwargs to accept multiple numbers and a
# keyword argument operation, and perform the specified operation such as sum, maximum, or
# minimum.
````

````python
def calculate(*args, **kwargs):
    operation = kwargs["operation"]

    if operation == "sum":
        result = sum(args)

    elif operation == "maximum":
        result = max(args)

    elif operation == "minimum":
        result = min(args)

    else:
        result = "Invalid operation"

    return result

operation=input("Enter the operation you want to perform:- ")
result = calculate(10, 20, 30, 40, operation=operation.lower())

print("Result:", result)
````

````python
# 21. Write a Python function using both *args and **kwargs to accept multiple items purchased and
# keyword arguments such as customer_name and discount, and calculate the final bill amount.
````

````python
def calculate_bill(*args, **kwargs):
    total = 0

    for price in args:
        total += price

    discount = kwargs["discount"]
    customer_name = kwargs["customer_name"]

    discount_amount = total * discount / 100
    final_amount = total - discount_amount

    print("Customer Name:", customer_name)
    print("Total Amount:", total)
    print("Discount:", discount, "%")
    print("Discount Amount:", discount_amount)
    print("Final Bill Amount:", final_amount)


calculate_bill(
    500, 1000, 2000,
    customer_name="Mohit",
    discount=10
)
````

````python
# 22. Write a Python function using both *args and **kwargs to accept multiple numbers and
# keyword arguments such as even or odd, and display the numbers according to the specified
# option.
````

````python
def display_numbers(*args, **kwargs):
    option = kwargs["option"]

    if option == "even":
        for number in args:
            if number % 2 == 0:
                print(number)

    elif option == "odd":
        for number in args:
            if number % 2 != 0:
                print(number)

    else:
        print("Invalid option")


option = input("Enter option (even/odd): ")

display_numbers(10, 15, 20, 25, 30, 35, option=option.lower())
````
