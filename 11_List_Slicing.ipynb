```python
lst=list(map(int, input("Enter the numbers in list:- ").split()))
```

```python
# Q1. Print list element from index 1 to 3
```

```python
result=lst[1:4]
result
```

```python
# Q2. print three elements of list:-
```

```python
result1=lst[:3]
result1
```

```python
# Q3 . print last three elements of list
```

```python
result3=lst[len(lst):-4:-1]
result3
```
