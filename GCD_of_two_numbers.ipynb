```python
x=int(input("enter the value of x: "))
y=int(input("enter the value of y: "))
```

```python
def gcd(a,b):
  if a<b:
    if a==0:
      return b
    else:
      return gcd(b%a, a)
  else:
    if b==0:
      return a
    else:
      return gcd(b,a%b)

```

```python
result=gcd(x,y)
print(result)
```
