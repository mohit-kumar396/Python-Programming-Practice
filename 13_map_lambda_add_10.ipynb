```python
lst=list(map(int, input("Enter teh value in list: ").split()))
```

```python
#Add 10 to each element of list using map() and lambda fn:-
# syntax :-
# list(map(lambda x: expression, iterable))
```

```python
resulted_list=list(map(lambda x:x+10, lst))
resulted_list
```

```python
#Replace every occurence of 15 by using map() and lambda fn:-
```

```python
lst1=[5,15,25,15,35]
result=list(map(lambda x:100 if x==15 else x, lst1))
result
```
