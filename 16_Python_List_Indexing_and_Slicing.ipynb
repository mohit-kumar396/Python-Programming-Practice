```python
# Q1. Python program to access and print the first, middle and last element of list.
```

```python
lst1=[10,20,30,40,50]
```

```python
for i in range(len(lst1)):
    if i==0:
        print(lst1[i], end=" ")
    elif i==len(lst1)//2:
        print(lst1[i], end=" ")
    elif i==len(lst1)-1:
        print(lst1[i], end=" ")
```

```python
# Q2. python program to print element in reverse order.
```

```python
lst1[: : -1]
```

```python
# concept of question 2
# start = blank
# stop  = blank
# step  = -1
# slicing me jab step = -1 hai then Python right side se iteration start karta hai, yani last element se.
# Stop bhi blank hai, iska matlab hai jitna possible ho utna continue karo.
# step = -1 har baar index ko 1 se decrease karo.
```

```python
# Q3.Python program to print only element at even indices.
```

```python
for i in range(len(lst1)):
  if i%2==0:
    print(lst1[i], end=" ")
```
